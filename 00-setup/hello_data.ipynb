This notebook created a orders array and then using dataframe which is a data type in pandas we create a tabular form of data using it then we perform varies operations like calculating and creating new column,sum,finding max,drawing chart,finding highest total,filtering,sorting this data


```python
orders = orders = [
    {"product": "Laptop",   "price": 55000, "qty": 2},
    {"product": "Mouse",    "price": 500,   "qty": 10},
    {"product": "Monitor",  "price": 12000, "qty": 3},
    {"product": "Keyboard", "price": 1500,  "qty": 5},
    {"product": "Headset",  "price": 2500,  "qty": 4},
]

print(orders[0])
print(len(orders))
```

```python
import pandas as pd
df = pd.DataFrame(orders)
print(df)
```

```python
df['total'] = df['price'] * df['qty']
df
```

```python
total_revenue = df['total'].sum()
print(f"Total Revenue: ₹{total_revenue}")

most_expensive = df.loc[df["price"].idxmax(),"product"]
print(f"Most expensive product: {most_expensive}")
```

```python
df.plot.bar(x="product",y="total",title="Revenue per product")
```

```python
highest_total = df.loc[df["total"].idxmax(),"product"]
print (f"Highest total revenue product: {highest_total}")
```

```python
filtered_df = df[df['qty'] > 3]
print("Products with quantity greater than 3:")
print(filtered_df)
```

```python
sorted_df = df.sort_values(by="total", ascending=False)
print("Products sorted by total revenue (descending):")
print(sorted_df)
```
